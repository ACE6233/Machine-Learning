```python
# Load modules and packages
import pandas as pd
from sklearn.model_selection import train_test_split as split
from sklearn.neighbors import KNeighborsClassifier as KNC
from sklearn.tree import DecisionTreeClassifier as DTC
```

Load the dataset

```python
df = pd.read_csv('https://raw.githubusercontent.com/wooihaw/datasets/main/fish.csv')
```

Check the dimension of the dataset

Examine the first few samples of the dataset

Goal: Given some fish samples, we want to classify their respective species.

Save the features in X, the label in y.

```python
X = df.drop(columns=["Species"])
y = df["Species"]
```

How many specifies are there?

Split the dataset into 2 portions for training and testing

```python
X_train, X_test, y_train, y_test = split(X, y, test_size=0.25, random_state=42)
```

Examine the dimensions of the training and testing datasets

How many data samples are there in each dataset?

How many features are used for training?

Train and evaluate a k-NN model with k=1

```python
knn = KNC(n_neighbors=1).fit(X_train, y_train)
print(f"KNN accuracy: {knn.score(X_test, y_test)}")
```

Train and evaluate a Decision Tree (DT) model with max_depth=1

```python
dtc = DTC(random_state=42, max_depth=1).fit(X_train, y_train)
print(f"DTC accuracy: {dtc.score(X_test, y_test)}")
```
